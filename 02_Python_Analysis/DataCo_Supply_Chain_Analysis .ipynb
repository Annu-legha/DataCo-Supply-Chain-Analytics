```python
# Create a copy of original dataset
df_clean = df.copy()

# Columns not required for business analysis
drop_columns = [
    'Product Description',
    'Order Zipcode',
    'Customer Email',
    'Customer Password',
    'Product Image',
    'Customer Street'
]

# Drop unnecessary columns
df_clean.drop(
    columns=drop_columns,
    inplace=True,
    errors='ignore'
)

print("Data cleaning completed!")

print("\nOriginal Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)

print("\nRemaining Missing Values:")
print(df_clean.isnull().sum().sort_values(ascending=False).head(10))

print("\nRemaining Columns:")
print(df_clean.columns.tolist())
```

```python
print("Final Dataset Information:")
df_clean.info()

print("\nFinal Missing Values:", df_clean.isnull().sum().sum())

print("\nDuplicate Rows:", df_clean.duplicated().sum())
```

```python
# Check date columns
date_cols = [
    col for col in df_clean.columns
    if 'date' in col.lower()
]

print("Date Columns:", date_cols)

# Check numerical columns
num_cols = df_clean.select_dtypes(
    include='number'
).columns.tolist()

print("\nNumerical Columns:", num_cols)
```

```python
date_cols = [
    'order date (DateOrders)',
    'shipping date (DateOrders)'
]

for col in date_cols:
    print("\nColumn:", col)
    print("Data Type:", df_clean[col].dtype)
    print("Sample Values:")
    print(df_clean[col].head())
```

```python
# Convert date columns into datetime format

date_cols = [
    'order date (DateOrders)',
    'shipping date (DateOrders)'
]

for col in date_cols:
    df_clean[col] = pd.to_datetime(
        df_clean[col],
        format='mixed',
        errors='coerce'
    )

print("Date conversion completed!")

print("\nUpdated Data Types:")
print(df_clean[date_cols].dtypes)

print("\nSample Dates:")
display(df_clean[date_cols].head())
```

```python
# Check invalid dates after conversion

for col in date_cols:
    print(
        col,
        "Invalid dates:",
        df_clean[col].isnull().sum()
    )

# Check date range
print("\nOrder Date Range:")
print(df_clean[date_cols[0]].min())
print(df_clean[date_cols[0]].max())

print("\nShipping Date Range:")
print(df_clean[date_cols[1]].min())
print(df_clean[date_cols[1]].max())
```

```python
# Feature Engineering

# Extract year, month and quarter
df_clean['Order_Year'] = df_clean['order date (DateOrders)'].dt.year

df_clean['Order_Month'] = df_clean['order date (DateOrders)'].dt.month

df_clean['Order_Month_Name'] = df_clean['order date (DateOrders)'].dt.month_name()

df_clean['Order_Quarter'] = df_clean['order date (DateOrders)'].dt.quarter

# Calculate actual delivery time
df_clean['Actual_Delivery_Days'] = (
    df_clean['shipping date (DateOrders)'] -
    df_clean['order date (DateOrders)']
).dt.days

# Calculate scheduled delivery time
df_clean['Scheduled_Delivery_Days'] = df_clean['Days for shipment (scheduled)']

# Calculate delivery delay
df_clean['Delivery_Delay_Days'] = (
    df_clean['Actual_Delivery_Days'] -
    df_clean['Scheduled_Delivery_Days']
)

# Classify delivery performance
df_clean['Delivery_Performance'] = df_clean['Delivery_Delay_Days'].apply(
    lambda x: 'Late' if x > 0 else 'On Time/Early'
)

print("Feature Engineering Completed!")

display(df_clean.head())
```

```python
print("New Columns:")
print([
    'Order_Year',
    'Order_Month',
    'Order_Month_Name',
    'Order_Quarter',
    'Actual_Delivery_Days',
    'Delivery_Delay_Days',
    'Delivery_Performance'
])

print("\nDelivery Performance Distribution:")
print(df_clean['Delivery_Performance'].value_counts())

print("\nDataset Shape:", df_clean.shape)
```

```python
# BUSINESS KPI ANALYSIS

total_orders = len(df_clean)

late_orders = (df_clean['Delivery_Performance'] == 'Late').sum()

on_time_orders = (df_clean['Delivery_Performance'] == 'On Time/Early').sum()

late_percentage = (late_orders / total_orders) * 100

on_time_percentage = (on_time_orders / total_orders) * 100

avg_delivery_days = df_clean['Actual_Delivery_Days'].mean()

avg_delay_days = df_clean['Delivery_Delay_Days'].mean()

avg_sales = df_clean['Sales'].mean() if 'Sales' in df_clean.columns else None

print("========== SUPPLY CHAIN KPIs ==========")

print("Total Orders:", total_orders)
print("Late Orders:", late_orders)
print("On-Time/Early Orders:", on_time_orders)

print(f"Late Delivery Rate: {late_percentage:.2f}%")
print(f"On-Time/Early Rate: {on_time_percentage:.2f}%")

print(f"Average Delivery Days: {avg_delivery_days:.2f}")
print(f"Average Delivery Delay: {avg_delay_days:.2f}")

if avg_sales is not None:
    print(f"Average Sales: {avg_sales:.2f}")
```

```python
# ===== FINAL DATE DATA QUALITY VALIDATION =====

date_comparison['Data_Quality_Status'] = np.where(
    date_comparison['Difference'].abs() <= 1,
    'Consistent',
    'Inconsistent'
)

# Count both categories
quality_summary = (
    date_comparison['Data_Quality_Status']
    .value_counts()
    .to_frame(name='Total_Records')
)

quality_summary['Percentage'] = (
    quality_summary['Total_Records'] /
    len(date_comparison) * 100
)

print("===== DATA QUALITY REPORT =====")

display(quality_summary)

print("\nTotal Records:", len(date_comparison))

print("Consistent Records:",
      (date_comparison['Data_Quality_Status'] == 'Consistent').sum())

print("Inconsistent Records:",
      (date_comparison['Data_Quality_Status'] == 'Inconsistent').sum())
```

```python
# ===== FINAL ANALYSIS DATASET =====

# Keep records with valid date sequences
df_final = df_analysis.loc[date_comparison.index].copy()

# Add data quality status
df_final['Date_Consistency_Status'] = (
    date_comparison['Data_Quality_Status']
)

# Calculate delivery delay using recorded shipping days
df_final['Delivery_Delay_Days'] = (
    df_final['Days for shipping (real)'] -
    df_final['Days for shipment (scheduled)']
)

# Classify delivery performance
df_final['Delivery_Performance'] = np.where(
    df_final['Delivery_Delay_Days'] > 0,
    'Late',
    'On Time/Early'
)

# Final KPIs
total_orders = len(df_final)

late_orders = (
    df_final['Delivery_Performance'] == 'Late'
).sum()

on_time_orders = total_orders - late_orders

print("===== FINAL SUPPLY CHAIN KPIs =====")

print("Total Orders:", total_orders)
print("Late Orders:", late_orders)
print("On-Time/Early Orders:", on_time_orders)

print("Late Delivery Rate:",
      round(late_orders / total_orders * 100, 2), "%")

print("On-Time/Early Rate:",
      round(on_time_orders / total_orders * 100, 2), "%")

print("Average Recorded Delivery Days:",
      round(df_final['Days for shipping (real)'].mean(), 2))

print("Average Delivery Delay:",
      round(df_final['Delivery_Delay_Days'].mean(), 2))

print("\nFinal Dataset Shape:", df_final.shape)
```
